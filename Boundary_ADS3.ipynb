```python
#Affine chart where multiple lines are intersecting one line going the other way on the plane
import numpy as np

var('x y z t p')

q = x^2 + y^2 - z^2 - 1

A = matrix([
    [cos(p), -sin(p), 0],
    [sin(p),  cos(p), 0],
    [0,       0,     1]
])

c = implicit_plot3d(
    q,
    (x, -3, 3),
    (y, -3, 3),
    (z, -2, 2),
    color='blue',
    opacity=0.3
)

line_1 = vector([1, t, t])
transformed_1 = A * line_1

p_values = np.arange(-10, 10.5, 0.5)


plots = []

for value in p_values:
    curve = transformed_1.subs(p=value)

    plots.append(
        parametric_plot3d(
            curve, (t, -2, 2),
            color='black',
            thickness=3
        )
    )


line_2 = vector([1, t, -t])
transformed_2 = A * line_2

for value in p_values:
    curve = transformed_2.subs(p=value)

    plots.append(
        parametric_plot3d(
            curve, (t, -2, 2),
            color='green',
            thickness=3
        )
    )


show(c + e + sum(plots), axes=True, aspect_ratio=1)
```
